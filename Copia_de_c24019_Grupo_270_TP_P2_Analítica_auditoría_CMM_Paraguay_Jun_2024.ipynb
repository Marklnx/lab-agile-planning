# Etapa 2: Exploración de los data sets
En SALES se observa que Brick y Steel tienen muchos valores 0 y/o de bajo valor comparados con los otros productos. Mientras que los valores de EXPORTS de esos mismos productos parecen normales comparados con el resto.
Pregunta: ¿en Paraguay no hay ventas de Brick y Steel?


```python
#VENTAS
suma_sales_py = df_sales_py[['Rubber', 'Brass', 'Vinyl', 'Granite', 'Stone', 'Brick', 'Aluminum', 'Glass', 'Plexiglass', 'Steel', 'Wood', 'Plastic']].sum()/1000000 #Expresada en millones

suma_sales_py #es una serie
```

```python
df_totales_sales_py = suma_sales_py.to_frame() #dataframe
df_totales_sales_py.reset_index(inplace=True)
df_totales_sales_py
```

```python
df_totales_sales_py.rename(columns={0:"Ventas_Totales (en MM)"}, inplace=True)
df_totales_sales_py.rename(columns={"index":"Producto"}, inplace=True)
df_totales_sales_py= df_totales_sales_py.sort_values(by='Ventas_Totales (en MM)', ascending=False)
print(df_totales_sales_py) #TOTAL VENTAS POR PRODUCTO
```

```python
#EXPORTACIONES
suma_exports_py = df_exports_py[['Rubber', 'Brass', 'Vinyl', 'Granite', 'Stone', 'Brick', 'Aluminum', 'Glass', 'Plexiglass', 'Steel', 'Wood', 'Plastic']].sum()/1000000 #Expresada en millones

suma_exports_py #es una serie
```

```python
df_totales_exports_py = suma_exports_py.to_frame() #dataframe
df_totales_exports_py.reset_index(inplace=True)
df_totales_exports_py
```

```python
df_totales_exports_py.rename(columns={0:"Exportaciones_Totales (en MM)"}, inplace=True)
df_totales_exports_py.rename(columns={"index":"Producto"}, inplace=True)
df_totales_exports_py= df_totales_exports_py.sort_values(by='Exportaciones_Totales (en MM)', ascending=False)
print(df_totales_exports_py) #TOTAL EXPORTACIONES POR PRODUCTO
```

```python
#MERGE ENTRE VENTAS Y EXPORTACIONES
df_merged = df_totales_sales_py.merge(df_totales_exports_py, how="inner", on="Producto")
pd.options.display.float_format = "$ {:,.2f}".format
df_merged
```

```python
df_merged['Ratio_Ventas/Exports'] = df_merged['Ventas_Totales (en MM)']/df_merged['Exportaciones_Totales (en MM)']
df_merged
```

```python
df_merged.describe()
```

# Etapa 3: Análisis Visual


```python
import matplotlib.pyplot as plt

ventas_totales = df_merged.groupby('Producto')['Ventas_Totales (en MM)'].mean()
exportaciones_totales = df_merged.groupby('Producto')['Exportaciones_Totales (en MM)'].mean()
productos = ventas_totales.index

fig, ax = plt.subplots(figsize=(12, 6))

#df_merged.groupby('Producto').mean().loc[:,['Ventas_Totales (en MM)','Exportaciones_Totales (en MM)']].plot(kind='bar', ax=ax)
ventas_totales.plot(kind='bar', ax=ax, color='green', label='Ventas Totales', width=0.3, position=1)
exportaciones_totales.plot(kind='bar', ax=ax, color='blue', label='Exportaciones Totales', width=0.3, position=0)

ax.set_title('Ventas y Exportaciones Totales por Producto')
ax.set_xlabel('Producto')
ax.set_ylabel('Valor (en MM)')
ax.set_xticklabels(productos, rotation=45)
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()
```

```python
total_exportaciones = sum(exportaciones_totales)
porcentajes = [100 * (ex / total_exportaciones) for ex in exportaciones_totales]

plt.figure(figsize=(5, 5))
plt.pie(porcentajes, labels=productos, autopct='%1.1f%%', startangle=140)
plt.title('Exportaciones Totales por Producto')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
total_ventas = sum(ventas_totales)
porcentajes = [100 * (ex / total_ventas) for ex in ventas_totales]

plt.figure(figsize=(5, 5))
plt.pie(porcentajes, labels=productos, autopct='%1.1f%%', startangle=140)
plt.title('Ventas Totales por Producto')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
plt.figure(figsize=(12, 6))  # Opcional: establecer el tamaño de la figura

plt.plot(df_merged['Producto'], df_merged['Ventas_Totales (en MM)'], "green", label='Ventas Totales (en MM)', marker='o')
plt.plot(df_merged['Producto'], df_merged['Exportaciones_Totales (en MM)'], 'blue' , label='Exportaciones Totales (en MM)', marker='p')

# Añadir títulos y etiquetas
plt.title('Comparación de Total Ventas y Total Exportaciones por Producto')
plt.xlabel('Producto')
plt.ylabel('Totales (en MM)')
plt.legend()  # Mostrar la leyenda

# Mostrar el gráfico
plt.show()
```

Claramente los Productos a reveer son Brick y Steel.

# **Análisis: conclusiones y oportunidades de mejora**

**Conclusiones**

Se identificó que los productos Steel y Brick no han generado ventas en el mercado paraguayo, lo que resulta en pérdidas por las exportaciones de estos productos. Por otro lado, Plexiglass ha demostrado ser el más vendido.

Razones detrás de las pérdidas:
La falta de ventas de Steel y Brick puede atribuirse a la inadecuación de estos materiales para las necesidades específicas del mercado paraguayo o la presencia de competidores locales que ofrecen opciones más económicas.

**Oportunidades de mejora**:




1 - Se recomienda aumentar las exportaciones de Plexiglass, ya que ha demostrado tener una demanda sólida al ser el producto más vendido.

2 - Auditar las exportaciones de Brick y Steel (los productos que no se han vendido) e intentar mitigar las pérdidas de futuras exportaciones de productos hacia nuevos mercados.

3 - Analizar en profundidad el mercado paraguayo para entender las necesidades y preferencias del público local antes de lanzar los productos. Teniendo en cuenta costos de producción, logística y estudiar bien a la competencia local.
